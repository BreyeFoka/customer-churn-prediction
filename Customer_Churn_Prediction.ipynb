## Identify Important Features

### Subtask:
Determine which features are most strongly correlated or associated with customer churn, using appropriate statistical methods or visualization.


**Reasoning**:
To enable correlation calculations and Chi-squared tests, the 'Churn' column needs to be converted into a numerical format. Then, I will convert 'Churn' to a numerical representation (1 for 'Yes', 0 for 'No') for the correlation calculations and Chi-squared tests.



````python
from scipy.stats import chi2_contingency

# Convert 'Churn' to numerical (0 for No, 1 for Yes)
df['Churn_Numeric'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)

print("--- Numerical Churn Column Created ---")
print(df[['Churn', 'Churn_Numeric']].head())
````

**Reasoning**:
Now that 'Churn' has been converted to a numerical column, I will calculate the Pearson correlation between each numerical feature and 'Churn_Numeric'. Additionally, I will perform Chi-squared tests for independence between each categorical feature and 'Churn_Numeric' to assess their association, and print the results.



````python
print("\n--- Pearson Correlation with Churn ---")
# Calculate correlation for numerical features
numerical_features_for_corr = [col for col in numerical_cols if col != 'Churn_Numeric'] # Ensure Churn_Numeric itself is not included

correlations = df[numerical_features_for_corr + ['Churn_Numeric']].corr()['Churn_Numeric'].drop('Churn_Numeric')
print(correlations)

print("\n--- Chi-squared Test for Categorical Features vs. Churn ---")
chi2_results = {}
# Add 'SeniorCitizen' to categorical_cols for chi-squared test as it's binary and often treated categorically
categorical_for_chi2 = categorical_cols_bivariate # This list already includes SeniorCitizen

for col in categorical_for_chi2:
    contingency_table = pd.crosstab(df[col], df['Churn'])
    chi2, p_value, _, _ = chi2_contingency(contingency_table)
    chi2_results[col] = {'chi2_statistic': chi2, 'p_value': p_value}

for col, results in chi2_results.items():
    print(f"Feature '{col}': Chi2-Statistic = {results['chi2_statistic']:.2f}, P-Value = {results['p_value']:.4f}")

print("\n--- Identifying Important Features ---")
print("Features with strong correlation (absolute value > 0.3) to Churn:")
strong_corr_features = correlations[abs(correlations) > 0.3]
if not strong_corr_features.empty:
    print(strong_corr_features)
else:
    print("No numerical features found with absolute correlation > 0.3.")

print("\nCategorical features with significant association (p-value < 0.05) to Churn:")
significant_chi2_features = {col: results for col, results in chi2_results.items() if results['p_value'] < 0.05}
if significant_chi2_features:
    for col, results in significant_chi2_features.items():
        print(f"- {col} (P-Value: {results['p_value']:.4f})")
else:
    print("No categorical features found with a p-value < 0.05.")
````

## Hyperparameter Tuning for Gradient Boosting with Class Weights

### Subtask:
Perform comprehensive hyperparameter tuning for a LightGBM model using RandomizedSearchCV. Focus on maximizing recall while explicitly handling class imbalance by incorporating class_weight or a similar strategy, to achieve a more accurate and robust model for churn prediction.


**Reasoning**:
To perform hyperparameter tuning for LightGBM and address class imbalance, I will first import the necessary libraries, calculate the `scale_pos_weight` based on the `y_train` distribution, define the parameter distributions for `RandomizedSearchCV`, initialize and fit `RandomizedSearchCV` with the LightGBM model, and then print the best parameters and recall score, finally storing the best estimator.



````python
import lightgbm as lgb
from sklearn.model_selection import RandomizedSearchCV
import numpy as np

print("--- Performing Hyperparameter Tuning for LightGBM ---")

# 1. Calculate scale_pos_weight for handling class imbalance
count_positive_samples = y_train.sum()
count_negative_samples = len(y_train) - count_positive_samples
scale_pos_weight = count_negative_samples / count_positive_samples
print(f"Calculated scale_pos_weight: {scale_pos_weight:.2f}")

# 2. Define a dictionary of parameter distributions (param_dist) for RandomizedSearchCV
param_dist = {
    'n_estimators': [100, 200, 300, 500, 700],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'num_leaves': [20, 31, 40, 50],
    'max_depth': [5, 7, 10, 12, -1], # -1 means no limit
    'min_child_samples': [20, 30, 40, 50],
    'subsample': [0.6, 0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.6, 0.7, 0.8, 0.9, 1.0],
    'reg_alpha': [0, 0.1, 0.5, 1.0],
    'reg_lambda': [0, 0.1, 0.5, 1.0]
}

# 3. Initialize RandomizedSearchCV
# Prioritize recall for churn prediction, use 5-fold cross-validation
# Pass scale_pos_weight to the LGBMClassifier
random_search = RandomizedSearchCV(estimator=lgb.LGBMClassifier(random_state=42, scale_pos_weight=scale_pos_weight),
                                   param_distributions=param_dist,
                                   n_iter=50, # Number of parameter settings that are sampled
                                   scoring='recall',
                                   cv=5,
                                   n_jobs=-1, # Use all available processors
                                   verbose=1,
                                   random_state=42)

# 4. Fit the RandomizedSearchCV object to the scaled training data
random_search.fit(X_train, y_train)

# 5. Print the best parameters and best score
print("\nBest Parameters for LightGBM:", random_search.best_params_)
print("Best Recall Score (LightGBM):", random_search.best_score_)

# 6. Store the best estimator (the optimized LightGBM model)
best_lgbm_model = random_search.best_estimator_
print("Optimized LightGBM model stored as 'best_lgbm_model'.")
````
